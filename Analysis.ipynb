# DATA CLEANING

```python
df = pd.read_csv(r"D:\final_project\Airline_project_01\Airline1.csv")
df.head(2)
```

```python
df.isnull().sum()
```

```python
df[df["Price"].isnull()]
mean_price = df["Price"].mean()
mean_price
df.fillna({'Price':mean_price},inplace=True)
```

```python
mode_Total_Stops = df['Total_Stops'].mode()[0]
mode_Total_Stops

df.fillna({'Total_Stops':mode_Total_Stops},inplace=True)

df[df['Total_Stops'].isnull()]
```

```python
mode_Desti = df['Destination'].mode()[0]

df.fillna({'Destination':mode_Desti},inplace=True)
df[df['Destination'].isnull()]
```

```python
df[df['Total_Stops'].isnull()]

mode_T_Stops =df['Total_Stops'].mode()[0]

df.fillna({'Total_Stops':mode_T_Stops},inplace=True)

df[df['Total_Stops'].isnull()]
```

```python
df[df.duplicated()]
```

```python
# remove
df.drop_duplicates(inplace=True)
```

# Graph Based Analysis

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
Air_count = df.groupby('Airline')['Airline'].count().sort_values()
Air_count.index
```

```python
# COUNT distribution of Airline.
ax = sns.countplot(x =df['Airline'],order=Air_count.index)
ax.bar_label(ax.containers[0])
plt.xticks(rotation=45)
plt.title("Count distribution of Airline")
plt.savefig("Presentation/1.CountdistributionofAirline.")
plt.show()
```

```python
# count Distribution of Source
Source_count = df.groupby('Source')['Source'].count().sort_values()
Source_count.index
```

```python
# Source count
ax = sns.countplot(x = df['Source'],order=Source_count.index)
ax.bar_label(ax.containers[0])
plt.title("Count Distribution of Source")
plt.savefig("Presentation/2.Sourcecount.png")

plt.show()
```

```python
# Price Distribution Against Airline
air_price = df.groupby('Airline')['Price'].sum().sort_values()
air_price.index
ax = sns.barplot(x=df['Airline'],y=df['Price'],estimator ='sum',order=air_price.index,errorbar=None)
ax.bar_label(ax.containers[0],rotation=90,fmt='%.0f')
plt.xticks(rotation=90)
plt.title(" Price Distribution Against Airline")
plt.savefig("Presentation/3.PricedistributionagainstAirline.png")

plt.show()
```

```python
# Price Against Source

Source_Price = df.groupby('Source')['Price'].sum().sort_values()
ax = sns.barplot(x=df['Source'],y=df['Price'],estimator='sum',errorbar=None,order = Source_Price.index)
ax.bar_label(ax.containers[0],rotation=80,fmt="%.0f")
plt.title("Price against Source")
plt.savefig("Presentation/4.PriceagainstSource.png")

plt.show()
```

```python
# Price against STOPS
stop_price = df.groupby('Total_Stops')['Price'].sum().sort_values()
ax = sns.barplot(x = df['Total_Stops'],y = df['Price'],estimator = 'sum',order =stop_price.index,errorbar=None)
ax.bar_label(ax.containers[0],rotation=90,fmt='%.0f')
plt.title("Price against stops")
plt.savefig("Presentation/5.PriceagainstSTOPS.png")
plt.show()
```

```python
# Avg Price count for toatal stops
Ttldf = df.groupby('Total_Stops')['Price'].agg(['mean','count']).sort_values(by ='mean')
Ttldf.index
ax = sns.barplot(Ttldf['mean'])
ax.twinx()
ax2 = sns.lineplot(Ttldf['count'],color ='red',marker='o')
ax.bar_label(ax.containers[0])
plt.title("Avg_price count for total stops")
plt.savefig("Presentation/6.Avg_price count for total stops.png")
plt.show()
```

```python
# AVG price count for Airline
# min price count for Airline
air_avg = df.groupby('Airline')['Price'].agg(['mean','count']).sort_values(by ='mean')
ax = sns.barplot(air_avg['mean'],order=air_avg.index)
plt.xticks(rotation=90)
ax.bar_label(ax.containers[0],rotation=85,fmt="%.0f")
ax.twinx()
ax2 = sns.lineplot(air_avg['count'],color='red',marker="X")
plt.title("Average price count for Airline")
plt.savefig("Presentation/7.Avg_pricecountforAirline.png")
plt.show()
```

```python
# min price count for Airline

air_min = df.groupby('Airline')['Price'].agg(['min','count']).sort_values(by ='min')
ax = sns.barplot(air_min['min'],order=air_min.index)
plt.xticks(rotation=90)
ax.bar_label(ax.containers[0],rotation=45,fmt="%.0f")
ax.twinx()
ax2 = sns.lineplot(air_min['count'],color='red',marker="X")
plt.title('Minimum price count for Airline')
plt.savefig("Presentation/8.minimumpricecountforAirline.png")
plt.show()
```

```python
df["Sector"] = df["Source"]+"-"+df["Destination"]

stops_df = df[df['Total_Stops']=='non-stop']
Sector_Df = stops_df.groupby('Sector')['Price'].agg('mean')

ax = sns.barplot(Sector_Df)
ax.bar_label(ax.containers[0])
plt.xticks(rotation=45)
plt.title('non-stop')
plt.savefig("Presentation/9.Totalstopsnon-stop.png")
plt.show()
```

```python
df["Sector"] = df["Source"]+"-"+df["Destination"]
list_Stops = enumerate(['non-stop','1 stop','2 stops','3 stops','4 stops'])
plt.figure(figsize=(15,20))
for index,value in list_Stops:
    stops_df = df[df['Total_Stops']==value]
    Sector_Df = stops_df.groupby('Sector')['Price'].agg('mean')
    plt.subplot(3,2,index+1)
    ax = sns.barplot(Sector_Df)
    ax.bar_label(ax.containers[0])
    plt.title(value)
plt.savefig("Presentation/10.avg_fareof eachSectorwithTotal-stops.png")
plt.show()
```
